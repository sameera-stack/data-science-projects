# Housing Data Analysis
This notebook explores the California housing dataset using Python, Pandas and Matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Loading the Data
The dataset has 17,000 rows and 9 columns, including median income, house age and median house value.

```python
df = pd.read_csv("/content/sample_data/california_housing_train.csv")
print(df.shape)
df.head()
```

## Checking the Data
I looked at summary statistics and missing values. All columns have 17,000 values and there are no missing values, so no cleaning was needed. The maximum house value is 500,001, which suggests that prices are capped at this level.

```python
print(df.describe())
print(df.isnull().sum())
```

## House Price Distribution
Most houses fall in the lower to middle price range, and the distribution is skewed to the right. There is also a spike at the highest value, which suggests that house prices in this dataset are capped at an upper limit.

```python
df["median_house_value"].hist(bins=40)
plt.title("Distribution of House Prices")
plt.xlabel("Median house value")
plt.ylabel("Count")
plt.show()
```

## Income vs House Value
Houses in areas with higher median income tend to have higher values, so the two are positively related. The horizontal line at the top is the price cap seen earlier, and it limits how high the values can go.

```python
df.plot.scatter(x="median_income", y="median_house_value", alpha=0.3)
plt.title("Income vs House Value")
plt.xlabel("Median income")
plt.ylabel("Median house value")
plt.show()
```

## Conclusion
Median income is positively related to house value in this dataset, although the price cap affects the highest values. Next, I plan to build a simple regression model to predict house prices.
